BAR PLOT

```python
lang=['c','java','python','c++']
student=[23,34,56,45]
plt.bar(lang,student)
plt.show()
```

```python
plt.bar([2,12,17,20,34],[14,20,25,26,30],label="marks",width=2)
plt.bar([2,4,13.20,24],[15.32,28,34,18],label="score",color='r',width=2)
plt.legend()
plt.xlabel('ranks')
plt.ylabel('percentage')
plt.title('student grade')
plt.show()
```

HISTOGRAMPLOT

```python
fig,ax=plt.subplots(1,1)
a=np.array([10,12,15,20,25,22])
ax.hist(a,bins=[10,12,13,22])
ax.set_title('my')
ax.set_xticks([10,15,20,30])
ax.set_ylabel('own')
ax.set_xlabel('gpa')
plt.show()
```

SCATTER

```python
plt.scatter([2,12,17,20,34],[14,20,25,26,30],label="marks")
plt.scatter([2,4,13.20,24],[15.32,28,34,18],label="score",color='r')
plt.legend()
plt.xlabel('ranks')
plt.ylabel('percentage')
plt.title('student grade')
plt.show()
```

VIOLIN PLOT

```python
marks=np.random.normal(100,10,150)
boys=np.random.normal(60,5,120)
girls=np.random.normal(80,8,130)

data=[marks,boys,girls]
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
m=ax.violinplot(data)
plt.show()
```
